```python
# @title Verbatim vs. Intended Transcription
# @markdown Sentence-Level Timestamps
# --------------------------------------------------------------
#   Whisper → segment‑level timestamps (one entry per sentence)
# --------------------------------------------------------------
import os
from IPython.display import Audio, display

# --------------------------------------------------------------
#   File handling (unchanged)
# --------------------------------------------------------------
UPLOADED_FILENAME = " Business-.mp3"      # @param {type:"string"}
audio_path = "/content/" + UPLOADED_FILENAME
if not os.path.exists(audio_path):
    audio_path = UPLOADED_FILENAME

# --------------------------------------------------------------
#   Transcribe with word‑level timestamps
# --------------------------------------------------------------
result = model.transcribe(
    audio_path,
    language="en",
    word_timestamps=True               # we need the words
)

# --------------------------------------------------------------
#   Helper: decide whether a word ends a sentence
# --------------------------------------------------------------
sentence_end_punct = {".", "!", "?"}
def ends_sentence(word: str) -> bool:
    # Strip surrounding punctuation and check the last char
    stripped = word.rstrip('\"\'”’')   # remove trailing quote‑like chars
    return stripped[-1] in sentence_end_punct if stripped else False

# --------------------------------------------------------------
#   Aggregate words → sentences
# --------------------------------------------------------------
sentences = []          # list of (start, end, text)
curr_words = []        # words of the sentence we are building
curr_start = None
curr_end   = None

for w in result.words:
    # w is a Whisper word object with .start, .end, .word
    if curr_start is None:
        curr_start = w.start                # first word of a new sentence

    curr_words.append(w.word)
    curr_end = w.end                        # keep moving the end forward

    if ends_sentence(w.word):
        # Sentence finished – store it
        sentence_text = " ".join(curr_words).replace(" .", ".")\
                                             .replace(" ,", ",")\
                                             .replace(" ?", "?")\
                                             .replace(" !", "!")
        sentences.append( (curr_start, curr_end, sentence_text) )
        # Reset for the next sentence
        curr_words = []
        curr_start = None
        curr_end   = None

# If the audio ends without a terminal punctuation, flush the remainder
if curr_words:
    sentence_text = " ".join(curr_words)
    sentences.append( (curr_start, curr_end, sentence_text) )

# --------------------------------------------------------------
#   Print the assembled sentence timestamps
# --------------------------------------------------------------
print("--- SENTENCE‑LEVEL TIMESTAMPS (built from words) ---")
for start, end, txt in sentences:
    print(f"{start:6.2f}s - {end:6.2f}s : {txt}")

print("…")
```
